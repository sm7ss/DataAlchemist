## ꒰🐚১ 2. Entrenamiento de Modelo por Región ໒꒱ིྀ

```python
from sklearn.metrics import root_mean_squared_error

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
random_state= config_geo0.ml_training.random_state

features= config_geo0.ml_preprocessing.columns
target= config_geo0.ml_training.target
test_size= config_geo0.ml_training.train_test_final

dict_pred= {}
```

```python
def train_model(x, y, name: str) -> Dict[str, Any]: 
    x_train, x_test, y_train, y_test= train_test_split(
        x, y, test_size=test_size, random_state=random_state
    )
    
    model= LinearRegression()
    model.fit(x_train, y_train)
    
    pred= model.predict(x_test)
    rmse= root_mean_squared_error(y_test, pred)
    
    print(f'RMSE {name}: {rmse}')
    return {
        'y_val': y_test, 
        'pred': pred, 
        'rmse': rmse, 
        'x_val': x_test
    }
```

```python
frames= [clean_frame_geo0, clean_frame_geo1, clean_frame_geo2]
names= ['geo0', 'geo1', 'geo2']

for i in range(len(frames)): 
    if names[i] == 'geo1': 
        frame= frames[i].select(['f0', 'f1', 'product'])
    else: 
        frame= frames[i].select(features)
    
    x= frame.drop(target).to_pandas(use_pyarrow_extension_array=True)
    y= frame[target].to_pandas(use_pyarrow_extension_array=True)
    
    pred= train_model(x=x, y=y, name=names[i])
    
    dict_pred[names[i]]= pred
    dict_pred[names[i]]['index']= x.iloc[pred['x_val'].index].index
```

## ⋆｡‧˚ʚ🧸 3. Preparación de Cálculo de Ganancias ɞ˚‧｡⋆

```python
unidad= 111.11
```

```python
def real_mean_calculo(y_real) -> str: 
    mean= sum(y_real)/len(y_real)
    redondeado= round(mean, 2)
    
    if mean > unidad: 
        return f'Se generaran ganacias puesto que el minimo de unidad es {unidad} y en este caso tenemos una media de {redondeado} unidades siendo {round(mean-unidad, 2)} más unidades en total a lo necesario'
    elif mean == unidad:
        return f'No se generaran ganancias puesto que lo mínimo necesario para no perder dinero == {unidad} unidades'
    else:
        return f'Se perderá dinero puesto que las unidades son menores a las necesarias {unidad} > {redondeado}'
```

```python
geo0_y= dict_pred['geo0']['y_val']
geo1_y= dict_pred['geo1']['y_val']
geo2_y= dict_pred['geo2']['y_val']

print('💰 Geo0 Ganacia')
print(real_mean_calculo(y_real=geo0_y))
print()
print('💰 Geo1 Ganancia')
print(real_mean_calculo(y_real=geo1_y))
print()
print('💰 Geo2 Ganacia')
print(real_mean_calculo(y_real=geo2_y))
```

```python
def create_frame(frame: pl.DataFrame, y_pred, y_real, index):
    id= frame.filter(pl.col('index').is_in(index)).get_column('id').to_list()
    
    dataframe= pl.DataFrame({
        'id': id, 
        'y_real': y_real, 
        'y_pred': y_pred
    })
    return dataframe
```

#### 💸 Ganancia 

```python
def calculo(name: str, frame: pl.DataFrame) -> None: 
    #Mejor caso
    suma= frame['y_pred'].sum()
    ingreso= suma*ingresos_u_vp
    
    diferiencia_de_inversion= ingreso - inversion
    conclusion_di= f'perdermos dinero {diferiencia_de_inversion:.3f}' if diferiencia_de_inversion < 0 else f'ganamos dinero; ~{diferiencia_de_inversion:.3f}'
    
    beneficio= diferiencia_de_inversion/total_pozos
    
    beneficio_prom_pozo= (beneficio/ingreso)*100
    margen_perdida_o_ganancia= 'perdida' if beneficio_prom_pozo < 0 else 'ganancia'
    
    #Teniendo en cuenta el error medio (rmse)
    rmse= dict_pred[name]['rmse']
    suma_con_rmse= suma - rmse
    ingresos_con_rmse= suma_con_rmse*ingresos_u_vp
    
    diferiencia_de_inversion_con_rmse= ingresos_con_rmse - inversion
    conclusion_di_rmse= f'perdermos dinero {diferiencia_de_inversion_con_rmse:.3f}' if diferiencia_de_inversion_con_rmse < 0 else f'ganamos dinero; ~{diferiencia_de_inversion_con_rmse:.3f}'
    
    beneficio_rmse= diferiencia_de_inversion_con_rmse/total_pozos
    
    beneficio_prom_rmse= (beneficio_rmse/ingresos_con_rmse)*100
    margen_perdida_o_ganancia_con_rmse= 'perdida' if beneficio_prom_rmse < 0 else 'ganancia'
    
    
    print(f'================================ 📑 {name} ===============================================')
    print('----------------------------------------------------------------------------------------')
    print('📈 Mejor escenario')
    print('----------------------------------------------------------------------------------------')
    print(f'📃 Suma total de volumen de petroleo en los 200 pozos con mejor volumen: {suma:.3f}')
    print(f'💵 Ingreso total de volumen de petroleo: {ingreso:.3f}')
    print(f'📍 Diferiencia de inversion e ingreso: {diferiencia_de_inversion:.3f} ({conclusion_di})')
    if diferiencia_de_inversion > 0: 
        print(f'    - 📈 Beneficio total: {beneficio:.3f}')
    else: 
        print(f'    - 📉 Perdida total: {beneficio:.3f}')
    
    print(f'📌 Margen: {beneficio_prom_pozo:.3f} ({margen_perdida_o_ganancia})')
    print('----------------------------------------------------------------------------------------')
    print('📉 Peor escenario incluyendo el error del modelo')
    print('----------------------------------------------------------------------------------------')
    print(f'📃 Suma total de volumen de petroleo en los 200 pozos con mejor volumen: {suma_con_rmse:.3f}')
    print(f'💵 Ingreso total de volumen de petroleo: {ingresos_con_rmse:.3f}')
    print(f'📍 Diferiencia de inversion e ingreso: {diferiencia_de_inversion_con_rmse:.3f} ({conclusion_di_rmse})')
    if diferiencia_de_inversion_con_rmse > 0: 
        print(f'    - 📈 Beneficio total: {beneficio_rmse:.3f}')
    else: 
        print(f'    - 📉 Perdida total: {beneficio_rmse:.3f}')
    
    print(f'📌 Margen: {beneficio_prom_rmse:.3f} ({margen_perdida_o_ganancia_con_rmse})')
    print()
```

## 🐱ྀི 5. Bootstrapping 𖾕𖾝꙼ᩚ𛲕𖾟

```python
riesgo_perdida= 2.5
muestras= 1000

fraction= 0.15
random_state=42

intervalo_de_confianza_min= 0.025
intervalo_de_confianza_max= 0.975
```

```python
def boostrapping(name: str, frame: pl.DataFrame) -> None:
    values= []
    
    for _ in range(muestras):
        sample= frame.sample(fraction=fraction, with_replacement=True, seed=random_state)
        sample_sorted= sample.sort('y_pred', descending=True)
        
        top_200= sample_sorted.limit(200)
        
        suma= top_200['y_real'].sum()
        beneficio= (suma*ingresos_u_vp)-inversion
        
        values.append(beneficio)
    
    valus= pl.Series(values)
    
    lower= valus.quantile(intervalo_de_confianza_min)
    upper= valus.quantile(intervalo_de_confianza_max)
    
    beneficio_prom= valus.mean()
    risk= ((valus < 0).sum() / len(valus))*100
    
    conclusion_risk= f'riesgo y fuera de rango permitido ({riesgo_perdida}%)' if risk > riesgo_perdida else f'fuera de riesgo, se puede invertir'
    
    print(f'========================== 📋 {name} ===========================')
    print('----------------------------------------------------------------')
    print(f'Ingresos mínimos ({intervalo_de_confianza_min}): ${lower:.3f}')
    print(f'Ingresos máximos ({intervalo_de_confianza_max}): ${upper:.3f}')
    print(f'Beneficio promedio: {beneficio_prom:.3f}')
    print(f'    - Conclusión: {conclusion_risk}. Risk percent: {risk:.3f}%')
    print('----------------------------------------------------------------')
    
    print()
```

```python
boostrapping(name='geo0', frame=geo0_preds)
boostrapping(name='geo1', frame=geo1_preds)
boostrapping(name='geo2', frame=geo2_preds)
```
