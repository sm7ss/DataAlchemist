## ꒰🐚১ Entrenamiento del modelo ໒꒱ིྀ

<div 
style="font-family: 'Segoe UI', Arial, sans-serif; 
white-space: pre-wrap; 
line-height: 1.2; 
font-size: 24px;">୨︶︶︶︶꒰᧔࿔᧓꒱︶︶︶︶୧୨︶︶︶︶꒰᧔࿔᧓꒱︶︶︶︶୧୨︶︶︶︶꒰᧔࿔᧓꒱︶︶︶︶୧୨︶︶︶︶꒰᧔࿔<br>
<b>⏔⏔⏔ ꒰ ᧔ 𝘉𝘶𝘴𝘲𝘶𝘦𝘥𝘢 𝘥𝘦 𝘮𝘦𝘫𝘰𝘳 𝘮𝘰𝘥𝘦𝘭𝘰 ᧓ ꒱ ⏔⏔⏔</b><br>
₊˚︶︶ꔫ︶︶‌｡˚₊˚︶︶ꔫ︶︶‌｡˚₊˚︶︶ꔫ︶︶‌｡˚₊˚︶︶ꔫ︶︶‌｡˚₊˚︶︶ꔫ︶︶‌｡˚₊˚︶︶ꔫ︶︶‌｡˚₊₊˚︶</div>

```python
model_hyperparameters= {
    'linear_regression': {
        'model': LinearRegression(fit_intercept=False),
        'params': {}
    },
    'random_forest_regressor': {
        'model': RandomForestRegressor(random_state=random_state),
        'params': {
            'n_estimators': [50, 100, 150],
            'max_depth': [5, 10], 
            'min_samples_split': [2, 5, 10]
        }
    },
    'decision_tree_regressor': {
        'model': DecisionTreeRegressor(random_state=random_state),
        'params': {
            'max_depth': [5, 10],
            'min_samples_split': [2, 5, 10]
        }
    },
    'xgb_regressor': {
        'model': XGBRegressor(random_state=random_state),
        'params': {
            'n_estimators': [100, 300],
            'learning_rate': [0.01, 0.05, 0.1],
            'max_depth': [3, 5, 7],
            'subsample': [0.8, 1.0],
            'colsample_bytree': [0.8, 1.0],
            'reg_lambda': [0.1, 1.0, 10.0],
            'reg_alpha': [0.0, 0.1, 1.0]
        }
    },
    'lbgm_regressor': {
        'model': LGBMRegressor(random_state=random_state),
        'params': {
            'n_estimators': [100, 300],
            'learning_rate': [0.01, 0.05, 0.1],
            'num_leaves': [15, 31, 63],
            'subsample': [0.8, 1.0],
            'colsample_bytree': [0.8, 1.0],
            'reg_lambda': [0.1, 1.0, 10.0],
            'reg_alpha': [0.0, 0.1, 1.0]
        }
    }
}
```

```python
for model_name, dict_model in model_hyperparameters.items(): 
    print(f"--- {model_name} ---")
    
    model= dict_model['model']
    params= dict_model['params']
    
    baseline= DummyRegressor(strategy='mean')
    baseline.fit(x_train_sample, y_train_sample)
    y_predbaseline= baseline.predict(x_test_sample)
    baseline_rmse= root_mean_squared_error(y_test_sample, y_predbaseline)
    
    # Overfitting
    if not params:
        model.fit(x_train_sample, y_train_sample)
        
        y_pred= model.predict(x_test_sample)
        model_rmse= root_mean_squared_error(y_test_sample, y_pred)
        
        s1= model.score(x_train_sample, y_train_sample)
        s2= model.score(x_test_sample, y_test_sample)
        
        gc.collect()
    else:
        randomized= RandomizedSearchCV(
            estimator=model, 
            param_distributions=params, 
            n_iter=20,
            scoring='neg_root_mean_squared_error',  
            n_jobs=n_jobs, 
            cv=3,
            random_state=random_state, 
            verbose=True
        )
        
        randomized.fit(x_train_sample, y_train_sample)
        y_pred= randomized.predict(x_test_sample)
        
        model_rmse= root_mean_squared_error(y_test_sample, y_pred)
        
        s1= randomized.score(x_train_sample, y_train_sample)
        s2= randomized.score(x_test_sample, y_test_sample)
        
        be= randomized.best_params_
        
        print(f'\n- ᨳ꒰ ୨ Best params ୧ ꒱ഒ -\n    {be}')
        
        gc.collect()
        print()
    
    print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ Resultados ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
    
    diff = s1 - s2
    
    if diff < baseline_rmse:  
        print('𝜗𝜚✩.    ✅ No hay overfitting significativo')
    else:
        print('𝜗𝜚✩.    ⚠️ Posible overfitting')
    print(f'•┈┈┈・・❥ Diferencia train-test RMSE: {diff:.3f}')
    
    # Underfitting
    if model_rmse < baseline_rmse:
        print('𝜗𝜚✩.    ✅ No hay underfitting (mejor que baseline)')
    else:
        print('𝜗𝜚✩.    ❌ Underfitting (peor que baseline)')
    print(f'•┈┈┈・・❥ Baseline RMSE: {baseline_rmse:.3f} - Modelo RMSE: {model_rmse:.3f}')
```
