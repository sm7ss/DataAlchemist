```python
high_corr= 0.90
estable_corr= config_var.correlation_decision_maker.threshold/100
```

```python
def corr_columns(
        frame: pl.DataFrame, 
        file: Dict[str, Any], 
    ) -> Dict[str, List[str]]: 
    columns= frame.columns
    
    high_corr_json= file['analysis_data']['correlation']['high_correlations']
    distribution_json= file['analysis_data']['distribution']
    
    eliminar_cols= []
    join_cols= []
    
    for i in range(len(high_corr_json)):
        puntos_corr= high_corr_json[i][-1]
        
        col_0= high_corr_json[i][0]
        col_1= high_corr_json[i][1]
        
        if puntos_corr >= high_corr: 
            eliminar_cols.append(col_1)
        elif estable_corr >= puntos_corr and high_corr < puntos_corr: 
            rango_col0= distribution_json[col_0]['range']
            rango_col1= distribution_json[col_1]['range']
            
            form_col0= distribution_json[col_0]['form']
            form_col1= distribution_json[col_1]['form']
            
            if form_col0 == form_col1: 
                if col_0 in eliminar_cols: 
                    continue
                else: 
                    join_cols.append([col_0, col_1])
                if col_1 in eliminar_cols: 
                    continue
                else: 
                    join_cols.append([col_0, col_1])
            else: 
                if rango_col0 > rango_col1: 
                    eliminar_cols.append(col_1)
                else: 
                    eliminar_cols.append(col_0)
        else: 
            continue
    
    for i in range(len(columns)): 
        col_test= columns[i]
        
        if 'final' in col_test.split('.'):
            if col_test == 'final.output.recovery': 
                continue
            eliminar_cols.append(col_test)
    
    return {
        'drop': eliminar_cols, 
        'join': join_cols
    }
```

```python
def corr_expr(frame: pl.DataFrame, cols_join: List[List[str]]) -> Dict[str, List[str]]: 
    list_expr= []
    list_col_drop= []
    
    for list_corr in range(len(cols_join)): 
        col_1= list_corr[0]
        col_2= list_corr[1]
        
        expr= CorrelationExpr(frame=frame).join(col_1=col_1, col_2=col_2)
        
        list_expr.append(expr)
        list_col_drop.append(col_1)
        list_col_drop.append(col_2)
    
    return {
        'drop': list_col_drop, 
        'expr': list_expr
    }
```

```python
def get_frame_preprocessed(
        pipeline_class: AutoPipeline,
        preprocessing_class: PreProcessinAuto,
        frame: pl.DataFrame,
        file: Dict[str, Any]
    ) -> pl.DataFrame: 
    corr= corr_columns(frame=frame, file=file)
    
    drop= corr['drop']
    join= corr['join']
    
    distribution= preprocessing_class.distribution() 
    outliers= preprocessing_class.outliers()
    
    frame= pipeline_class.nulls_frame(frame=frame)
    
    if distribution: 
        frame= frame.with_columns(distribution)
    if outliers is not None: 
        frame= outliers
    if join: 
        corr= corr_expr(frame=frame, cols_join=join)
        
        drop_corr= corr['drop']
        expr= corr['expr']
        
        if drop_corr: 
            drop+= drop_corr
            drop= tuple(drop)
        if expr: 
            frame= frame.with_columns(expr)
    
    if drop: 
        frame= frame.drop(drop)
    
    return frame.drop('index').fill_null(0).fill_nan(0)
```

```python
random_state= config.ml_training.random_state
mh= config_modeling.models_hyperparameters

target= ['rougher.output.recovery', 'final.output.recovery']
test_size= config.ml_training.train_test_search

cv= config_modeling.grid_search_cv.cv
n_jobs= config_modeling.grid_search_cv.n_jobs - 7
```

```python
overfitting= 10
```

```python
scaler_train= SelectScaler(
    frame=train_preprocessed, 
    config_ml=config_modeling
).auto()
scaler_test= SelectScaler(
    frame=test_preprocessed, 
    config_ml=config_test_modeling
).auto()
```

```python
test_columns= test_preprocessed.columns
train_columns= train_preprocessed.columns

train_cols= []
test_cols= []

for col in test_columns: 
    for col_t in train_columns: 
        if col in train_columns and col_t in test_columns: 
            train_cols.append(col)
            test_cols.append(col)
            break

train_cols+= target
```

```python
test_preprocessed= test_preprocessed.select(test_cols)
train_preprocessed= train_preprocessed.select(train_cols)
```

```python
test_columns= test_preprocessed.columns
train_columns= test_columns + target

train_preprocessed= train_preprocessed.select(train_columns)
```

```python
model_hyperparameters= {
    'decision_tree': {
        'model': DecisionTreeRegressor(random_state=random_state), 
        'params': {
            'max_depth': mh.decision_tree.max_depth, 
            'min_samples_split': mh.decision_tree.min_samples_split
        }
    },
    'random_forest': {
        'model': RandomForestRegressor(random_state=random_state), 
        'params': { 
            'n_estimators': [50, 100, 200], 
            'max_depth': [5, 10], 
            'min_samples_split': [2, 5, 10]
        }
    },
    'linear_regression' :{ 
        'model': LinearRegression(),
        'params': { 
            'fit_intercept': mh.linear_regression.fit_intercept, 
        }
    }
}

scoring_gs_cv= [
    'neg_mean_squared_error',
    'neg_root_mean_squared_error',
    'r2',
]
```

### .✩𝜗𝜚 ⌒⌒⌒ 𝘉ú𝘴𝘲𝘶𝘦𝘥𝘢 𝘥𝘦 𝘮𝘦𝘫𝘰𝘳𝘦𝘴 𝘩𝘪𝘱𝘦𝘳𝘱𝘢𝘳á𝘮𝘦𝘵𝘳𝘰𝘴 𝘺 𝘮𝘰𝘥𝘦𝘭𝘰 ⌒⌒⌒ 𝜗𝜚✩.

```python
sample_train= SamplingData(
    frame=train_preprocessed, 
    config=config_preprocessing
).sampling(decision='random')
sample_test= SamplingData(
    frame=test_preprocessed, 
    config=config_test_preprocessing
).sampling(decision='random')
```

```python
target_rough= sample_train[target[0]].to_pandas(use_pyarrow_extension_array=True)
target_final= sample_train[target[1]].to_pandas(use_pyarrow_extension_array=True)
```

```python
# Rougher
columns_rougher= []
for col in sample_train.columns: 
    if 'rougher' in col.split('.'): 
        columns_rougher.append(col)

x_rougher= sample_train.select(columns_rougher).drop(target, strict=False).to_pandas(use_pyarrow_extension_array=True)

columns_rougher= [col for col in sample_test.columns if 'rougher' in col.split('.')]
x_final_rough= sample_test.select(columns_rougher).drop(target, strict=False).to_pandas(use_pyarrow_extension_array=True)
```

```python
# Final
x_final= sample_train.drop(target, strict=False).to_pandas(use_pyarrow_extension_array=True)
x_final_test= sample_test.drop(target, strict=False).to_pandas(use_pyarrow_extension_array=True)
```

```python
if scaler_train: 
    x_rougher= scaler_train.fit_transform(x_rougher)
    x_final_rough= scaler_train.transform(x_final_rough) 
```

```python
if scaler_test: 
    x_final= scaler_test.fit_transform(x_final)
    x_final_test= scaler_test.transform(x_final_test)
```

#### 𐔌՞. .՞𐦯────୨ 𝘙𝘰𝘶𝘨𝘩𝘦𝘳 ৎ────⊹

```python
x_train, x_search, y_train, y_search= train_test_split(
    x_rougher, target_rough, test_size=test_size, random_state=random_state, shuffle=True
)
```

```python
for model_name, dict_model in model_hyperparameters.items(): 
    print(f"--- {model_name} ---")
    
    model= dict_model['model']
    params= dict_model['params']
    
    baseline= DummyRegressor(strategy='mean')
    baseline.fit(x_train, y_train)
    y_predbaseline= baseline.predict(x_search)
    baseline_acc= mean_squared_error(y_search, y_predbaseline)
    
    for scoring in scoring_gs_cv:
        grid_search= GridSearchCV(
            model, 
            params, 
            cv=cv, 
            scoring=scoring,  
            n_jobs=n_jobs
        )
        
        # Rendimiento del modelo por dentro
        grid_search.fit(x_train, y_train)
        y_pred= grid_search.predict(x_search)
        
        if scoring == 'neg_mean_squared_error': 
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ Mean Squared error ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            s1= grid_search.score(x_train, y_train) 
            s2= grid_search.score(x_search, y_search)
            acc= mean_squared_error(y_search, y_pred)
            
            difference= s1 - s2
            
            if difference < overfitting: 
                print('𝜗𝜚✩.    El modelo no tiene overfitting')
            else: 
                print('𝜗𝜚✩.    El modelo tiene overfitting')
            print(f'•┈┈┈・・❥ La diferencia entre train y test es de: {difference:.3f}')
            
            if acc > baseline_acc: 
                print('𝜗𝜚✩.    El modelo no tiene underfitting')
            else: 
                print('𝜗𝜚✩.    El modelo posee underfitting, es decir, es peor que el modelo "tonto" o de "piso"')
            print(f'•┈┈┈・・❥ El modelo de piso tiene un mse de {baseline_acc:.3f}; mientras que el mse de predicción tiene un mse de {acc:.3f}')
        elif scoring == 'neg_root_mean_squared_error':
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ Root Mean Squared Error ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            ps= root_mean_squared_error(y_search, y_pred)
            print(f'•┈┈┈・・❥ RMSE: {ps:.3f}')
        else: 
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ R2 Score ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            r2= r2_score(y_search, y_pred)
            print(f'•┈┈┈・・❥ R2 Score: {r2:.3f}')
    
    be= grid_search.best_params_
    print(f'\n- ᨳ꒰ ୨ Best params ୧ ꒱ഒ -\n    {be}')
    
    print()
```

```python
x_train, x_search, y_train, y_search= train_test_split(
    x_final, target_final, test_size=test_size, random_state=random_state, shuffle=True
)
```

```python
for model_name, dict_model in model_hyperparameters.items(): 
    print(f"--- {model_name} ---")
    
    model= dict_model['model']
    params= dict_model['params']
    
    baseline= DummyRegressor(strategy='mean')
    baseline.fit(x_train, y_train)
    y_predbaseline= baseline.predict(x_search)
    baseline_acc= mean_squared_error(y_search, y_predbaseline)
    
    for scoring in scoring_gs_cv:
        grid_search= GridSearchCV(
            model, 
            params, 
            cv=cv, 
            scoring=scoring,  
            n_jobs=n_jobs
        )
        
        # Rendimiento del modelo por dentro
        grid_search.fit(x_train, y_train)
        y_pred= grid_search.predict(x_search)
        
        if scoring == 'neg_mean_squared_error': 
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ Mean Squared error ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            s1= grid_search.score(x_train, y_train) 
            s2= grid_search.score(x_search, y_search)
            acc= mean_squared_error(y_search, y_pred)
            
            difference= s1 - s2
            
            if difference < overfitting: 
                print('𝜗𝜚✩.    El modelo no tiene overfitting')
            else: 
                print('𝜗𝜚✩.    El modelo tiene overfitting')
            print(f'•┈┈┈・・❥ La diferencia entre train y test es de: {difference:.3f}')
            
            if acc > baseline_acc: 
                print('𝜗𝜚✩.    El modelo no tiene underfitting')
            else: 
                print('𝜗𝜚✩.    El modelo posee underfitting, es decir, es peor que el modelo "tonto" o de "piso"')
            print(f'•┈┈┈・・❥ El modelo de piso tiene un mse de {baseline_acc:.3f}; mientras que el mse de predicción tiene un mse de {acc:.3f}')
        elif scoring == 'neg_root_mean_squared_error':
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ Root Mean Squared Error ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            ps= root_mean_squared_error(y_search, y_pred)
            print(f'•┈┈┈・・❥ RMSE: {ps:.3f}')
        else: 
            print('\n໒꒱ ‧₊˚ ྀིა₊ ⊹  ᧔ R2 Score ᧓  ໒꒱ ‧₊˚ ྀིა₊ ⊹')
            
            r2= r2_score(y_search, y_pred)
            print(f'•┈┈┈・・❥ R2 Score: {r2:.3f}')
    
    be= grid_search.best_params_
    print(f'\n- ᨳ꒰ ୨ Best params ୧ ꒱ഒ -\n    {be}')
    
    print()
```

```python
def sMAPE(y, y_pred) -> float: 
    dif= abs(y - y_pred)
    div= (abs(y) + abs(y_pred)) 
    
    diff= 2*dif/div
    mean= np.mean(diff)
    
    return mean*100
```
